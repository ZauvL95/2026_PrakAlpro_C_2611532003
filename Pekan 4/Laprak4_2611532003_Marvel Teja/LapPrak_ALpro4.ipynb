## 2.1 Dasar Teori

Percabangan merupakan salah satu konsep dasar dalam pemrograman yang digunakan untuk membuat program mengambil keputusan berdasarkan kondisi tertentu. Dengan percabangan, program dapat menjalankan perintah yang berbeda sesuai dengan hasil dari suatu kondisi. Dalam bahasa pemrograman Python, percabangan dapat dilakukan menggunakan beberapa struktur, seperti if, if-else, if-elif-else, multi-if, dan match-case.

1. If
    if digunakan untuk menjalankan suatu perintah apabila kondisi yang diberikan bernilai benar (True). Jika kondisi bernilai salah (False), perintah di dalam if tidak dijalankan.
2. If-Else
    if-else digunakan ketika terdapat dua kemungkinan kondisi. Jika kondisi pada if bernilai benar, perintah pada if dijalankan. Jika kondisi salah, perintah pada else akan dijalankan.
3. If-Elif-Else
    Struktur if-elif-else digunakan untuk memeriksa beberapa kondisi secara berurutan. Python akan menjalankan blok perintah pertama yang kondisinya bernilai benar. Jika tidak ada kondisi yang terpenuhi, maka blok else akan dijalankan.
4. Multi-If
    Multi-if merupakan penggunaan beberapa pernyataan if yang berdiri secara terpisah. Setiap kondisi akan diperiksa secara independen, sehingga lebih dari satu kondisi dapat terpenuhi dan dijalankan dalam satu program.
5. Match-Case
    match-case digunakan untuk mencocokkan suatu nilai dengan beberapa pilihan atau pola tertentu. Struktur ini dapat digunakan sebagai alternatif ketika terdapat banyak pilihan berdasarkan satu nilai, sehingga kode menjadi lebih terstruktur.
6. Operator Perbandingan dan Logika
    Percabangan menggunakan operator perbandingan seperti >, <, >=, <=, ==, dan != untuk membandingkan nilai. Selain itu, operator logika seperti and, or, dan not dapat digunakan untuk menggabungkan atau membalik kondisi.

### 2.2.4 Program "Match_Case1_2611532003.py"

```python
# Buat file dengan nama Match_case1_NIM.py
# Buat program untuk match case
# Nama variable ditambah 4 digit nim terakhir contoh: ipk_1234
# Program ini mengggunakan fungsi input()
# Program konversi angka menjadi nama bulan

bulan_2003 = int(input("Masukkan angka bulan (1-12): "))

match bulan_2003:
    case 1:
        print("Januari")
    case 2:
        print("Febuari")
    case 3:
        print("Maret")
    case 4:
        print("April")
    case 5:
        print("Mei")
    case 6:
        print("Juni")
    case 7:
        print("Juli")
    case 8:
        print("Agustus")
    case 9:
        print("September")
    case 10:
        print("Oktober")
    case 11:
        print("November")
    case 12:
        print("Desember")
    case _:
        print("Angka tidak valid")
```

**Penjelasan :**
1. Bagian awal berfungsi untuk menerima input angka bulan dari pengguna menggunakan fungsi input() lalu mengonversinya menjadi tipe data int yang disimpan dalam variabel bulan_2003.
2. Bagian match berfungsi untuk mencocokkan nilai bulan_2003 dengan beberapa pilihan angka yang tersedia, yaitu angka 1 sampai 12.
3. Bagian case 1 sampai case 12 berfungsi untuk menentukan nama bulan berdasarkan angka yang dimasukkan pengguna. Contohnya, jika pengguna memasukkan angka 1, program akan menampilkan "Januari".
4. Bagian case _ berfungsi sebagai kondisi default yang dijalankan jika angka yang dimasukkan tidak sesuai dengan angka 1 sampai 12. Program akan menampilkan pesan "Angka tidak valid".
5. Program ini menggunakan struktur match-case sebagai alternatif percabangan untuk memilih satu kondisi berdasarkan nilai yang diberikan.

### 2.2.6 Program "Multi_if2_2611532003.py"

```python
# Buat file dengan nama Multi_if2_NIM.py
# Buat program untuk kondisional if
# Nama variable ditambah 4 digit nim terakhir contoh: total_belanja_NIM.py
# Program ini mengggunakan fungsi input()
# Program Menghitung Diskon Belanja

# Input dari user
total_belanja_2611532003 = float(input("Masukkan total belanja (Rp): "))

# Input status member (mengecek apakah user mengetik 'y' atau 'ya')
input_member_2003 = input("Apakah anda member? (y/t): ").strip().lower()
is_member_2003 = input_member_2003 in ['y', "ya"]

# Input status kode promo (mengecek apakah user mengetik 'y' atau 'ya')
input_promo_2003 = input("Apakah kode promo valid? (y/t): ").strip().lower()
kode_promot_valid_2003 = input_promo_2003 in ['y', "y"]

total_diskon_persen_2003 = 0

# Multi-IF terpisah: setiap kondisi diperiksa secara independen
# Diskon bisa ditumpuk  (akumulasi) jika memenuhi beberapa syarat sekaligus

if total_belanja_2611532003 > 1000000:
    total_diskon_persen_2003 += 10

if is_member_2003:
    total_diskon_persen_2003 += 5

if kode_promot_valid_2003:
    total_diskon_persen_2003 += 15

# Menghitung nominal  diskon dan total belanja
nominal_diskon_2003 = total_belanja_2611532003 * (total_diskon_persen_2003 / 100)
total_bayar_2003 = total_belanja_2611532003 - nominal_diskon_2003

# Output hasil
print("/n--- Rincian Pembayaran ---")
print(f"Total Diskon    : {total_diskon_persen_2003}%(Rp{nominal_diskon_2003:,.0f})")
print(f"Total Bayar     : Rp {total_bayar_2003: ,.0f}")

print(f"Total diskon yang anda dapatkan: {total_diskon_persen_2003}%")
# Output: Total diskon yang anda dapatkan: 30% jika belanja > 1 juta, member, dan kode promo valid
```

**Penjelasan :**
1. Bagian awal berfungsi untuk menerima input total belanja dari pengguna menggunakan fungsi input() lalu mengonversinya menjadi tipe data float yang disimpan dalam variabel total_belanja_2611532003.
2. Bagian input status member berfungsi untuk menerima jawaban pengguna mengenai keanggotaan. Fungsi strip() digunakan untuk menghapus spasi di awal dan akhir input, sedangkan lower() digunakan untuk mengubah input menjadi huruf kecil. Hasilnya disimpan dalam variabel is_member_2003 dengan nilai True atau False.
3. Bagian input kode promo berfungsi untuk menerima status kode promo dari pengguna. Input diperiksa menggunakan operator in untuk menentukan apakah jawaban sesuai dengan pilihan yang ditentukan. Hasilnya disimpan dalam variabel kode_promot_valid_2003.
4. Bagian total_diskon_persen_2003 = 0 berfungsi untuk memberikan nilai awal diskon sebesar 0% sebelum program memeriksa kondisi-kondisi diskon.
5. Bagian if pertama berfungsi untuk mengecek apakah total belanja lebih dari Rp1.000.000. Jika terpenuhi, program menambahkan diskon sebesar 10%.
6. Bagian if kedua berfungsi untuk mengecek apakah pengguna merupakan member. Jika bernilai True, program menambahkan diskon sebesar 5%.
7. Bagian if ketiga berfungsi untuk mengecek apakah kode promo yang dimasukkan valid. Jika bernilai True, program menambahkan diskon sebesar 15%.
8. Ketiga kondisi tersebut menggunakan multi-if, sehingga setiap kondisi diperiksa secara independen dan diskon dapat dijumlahkan jika beberapa kondisi terpenuhi. Jika semua kondisi terpenuhi, total diskon yang diperoleh adalah 30%.
9. Bagian perhitungan berfungsi untuk menghitung nominal diskon dengan mengalikan total belanja dengan persentase diskon. Setelah itu, total pembayaran dihitung dengan mengurangi total belanja dengan nominal diskon.
10. Bagian output berfungsi untuk menampilkan rincian pembayaran, yaitu persentase diskon, nominal diskon, total pembayaran, dan total diskon yang diperoleh pengguna.

```python
print("=== SISTEM LOKET ALPRO ADVENTURE PARK ===")

# ------------------------------------------------------------
# 1. INPUT DATA PENGUNJUNG
# ------------------------------------------------------------

nama_2003 = input("Masukkan Nama Pengunjung        : ")
umur_2003 = int(input("Input umur anda                 : "))
sim_2003 = input(
    "Apakah Anda Sudah Punya SIM C (y/t): "
).strip().lower()[0]

print("\nPilihan Paket Wahana (1-5):")
print(" 1. Safari Rimba         (Rp 50,000)")
print(" 2. Arung Jeram          (Rp 75,000)")
print(" 3. Motor ATV Ekstrim    (Rp 120,000)")
print(" 4. Roller Coaster Kilat (Rp 100,000)")
print(" 5. All-Access VIP       (Rp 220,000)")

paket_2003 = int(input("\nMasukkan nomor paket (1-5)   : "))
jumlah_tiket_2003 = int(input("Masukkan jumlah tiket         : "))
is_member_2003 = input("Apakah Anda member? (y/t)     : ").strip().lower()
kode_promo_valid_2003 = input("Apakah kode promo valid? (y/t): ").strip().lower()


# ------------------------------------------------------------
# 2. VALIDASI JUMLAH TIKET
# IF TUNGGAL
# ------------------------------------------------------------

if jumlah_tiket_2003 <= 0:
    print("\nPeringatan: Kuota tiket tidak valid.")


# ------------------------------------------------------------
# 3. PEMILIHAN WAHANA DENGAN MATCH-CASE
# ------------------------------------------------------------

match paket_2003:

    case 1:
        nama_wahana_2003 = "Wahana Safari Rimba"
        harga_satuan_2003 = 50000

    case 2:
        nama_wahana_2003 = "Wahana Arung Jeram"
        harga_satuan_2003 = 75000

    case 3:
        nama_wahana_2003 = "Wahana Motor ATV Ekstrim"
        harga_satuan_2003 = 120000

    case 4:
        nama_wahana_2003 = "Wahana Roller Coaster Kilat"
        harga_satuan_2003 = 100000

    case 5:
        nama_wahana_2003 = "Wahana All-Access VIP"
        harga_satuan_2003 = 220000

    case _:
        print("\nPaket wahana tidak valid!")
        raise SystemExit


# ------------------------------------------------------------
# 4. VALIDASI KELAYAKAN PENGENDARA
# TIDAK MENGGUNAKAN NESTED-IF
# ------------------------------------------------------------

print("\n--- KELAYAKAN PENGENDARA WAHANA ---")

if paket_2003 == 3 and umur_2003 >= 17 and sim_2003 == 'y':
    status_akses_2003 = (
        "Anda sudah dewasa dan boleh mengendarai ATV sendiri."
    )

elif paket_2003 == 3 and umur_2003 >= 17 and sim_2003 != 'y':
    status_akses_2003 = (
        "Anda sudah dewasa tetapi tidak boleh bawa motor ATV "
        "(wajib didampingi instruktur)."
    )

elif paket_2003 == 3 and umur_2003 < 17 and sim_2003 == 'y':
    status_akses_2003 = (
        "Identitas tidak valid: Belum cukup umur memiliki SIM."
    )

elif paket_2003 == 3 and umur_2003 < 17 and sim_2003 != 'y':
    status_akses_2003 = (
        "Anda belum cukup umur dan tidak boleh bawa motor ATV."
    )

elif paket_2003 != 3 and umur_2003 >= 10:
    status_akses_2003 = (
        "Anda memenuhi batas usia untuk wahana ini."
    )

else:
    status_akses_2003 = (
        "Anda belum cukup umur untuk wahana ini."
    )

print("Status Akses:", status_akses_2003)


# ------------------------------------------------------------
# 5. MENGHITUNG SUBTOTAL
# ------------------------------------------------------------

subtotal_2003 = harga_satuan_2003 * jumlah_tiket_2003


# ------------------------------------------------------------
# 6. MULTI-IF TERPISAH UNTUK DISKON AKUMULATIF
# ------------------------------------------------------------

total_diskon_persen_2003 = 0

# Diskon Belanja Besar
if subtotal_2003 >= 200000:
    total_diskon_persen_2003 += 10

# Diskon Member
if is_member_2003 in ['y', 'ya']:
    total_diskon_persen_2003 += 5

# Diskon Voucher Promo
if kode_promo_valid_2003 in ['y', 'ya']:
    total_diskon_persen_2003 += 15

# Diskon Tambahan Rombongan
if jumlah_tiket_2003 >= 5:
    total_diskon_persen_2003 += 5


# ------------------------------------------------------------
# 7. MENGHITUNG TOTAL PEMBAYARAN
# ------------------------------------------------------------

nominal_diskon_2003 = (
    subtotal_2003 * (total_diskon_persen_2003 / 100)
)

total_bayar_2003 = (
    subtotal_2003 - nominal_diskon_2003
)


# ------------------------------------------------------------
# 8. EVALUASI BONUS
# IF-ELSE
# ------------------------------------------------------------

if total_bayar_2003 > 300000:
    catatan_layanan_2003 = (
        "Selamat! Anda berhak mendapatkan Souvenir Gratis."
    )
else:
    catatan_layanan_2003 = (
        "Terima kasih telah berkunjung."
    )


# ------------------------------------------------------------
# 9. MENAMPILKAN RINCIAN PEMBAYARAN
# ------------------------------------------------------------

print(f"Subtotal Belanja  : Rp {subtotal_2003:,.0f}")
print(
    f"Total Diskon      : {total_diskon_persen_2003}% "
    f"(Rp {nominal_diskon_2003:,.0f})"
)
print(f"Total Bayar       : Rp {total_bayar_2003:,.0f}")
print(f"Catatan Layanan   : {catatan_layanan_2003}")
print("\nProgram Selesai")
```

**Penjelasan :**

1. Bagian awal berfungsi untuk menampilkan judul program dan menerima data pengunjung berupa nama, umur, kepemilikan SIM C, pilihan paket wahana, jumlah tiket, status member, dan kode promo menggunakan fungsi input().
2. Bagian validasi jumlah tiket menggunakan if tunggal untuk mengecek apakah jumlah tiket kurang dari atau sama dengan 0. Jika kondisi terpenuhi, program menampilkan peringatan bahwa jumlah tiket tidak valid.
3. Bagian pemilihan wahana menggunakan struktur match-case untuk menentukan nama wahana dan harga satuan berdasarkan nomor paket yang dipilih. Jika nomor yang dimasukkan tidak tersedia, program menampilkan pesan bahwa paket tidak valid dan menghentikan program.
4. Bagian kelayakan pengendara menggunakan struktur if-elif-else untuk menentukan apakah pengunjung memenuhi syarat menggunakan wahana. Untuk paket ATV, program mengecek umur dan kepemilikan SIM C. Untuk wahana lainnya, program mengecek batas usia minimal 10 tahun. Bagian ini tidak menggunakan nested-if karena setiap kondisi ditulis dalam satu tingkat percabangan.
5. Bagian perhitungan subtotal berfungsi untuk menghitung total harga tiket dengan mengalikan harga satuan wahana dengan jumlah tiket menggunakan operator perkalian *.
6. Bagian pemberian diskon menggunakan beberapa if terpisah (multi-if). Diskon diberikan berdasarkan subtotal belanja, status member, kode promo yang valid, dan jumlah tiket. Setiap diskon yang memenuhi syarat akan ditambahkan sehingga diskon dapat terakumulasi.
7. Bagian perhitungan total pembayaran berfungsi untuk menghitung nominal diskon berdasarkan persentase diskon yang diperoleh, kemudian mengurangkannya dari subtotal untuk mendapatkan total yang harus dibayar.
8. Bagian evaluasi bonus menggunakan struktur if-else untuk mengecek apakah total pembayaran lebih dari Rp300.000. Jika terpenuhi, pengunjung mendapatkan souvenir gratis. Jika tidak, program menampilkan ucapan terima kasih.
9. Bagian akhir berfungsi untuk menampilkan rincian pembayaran berupa subtotal belanja, total persentase diskon, nominal diskon, total pembayaran, dan catatan layanan, kemudian menampilkan pesan "Program Selesai".
