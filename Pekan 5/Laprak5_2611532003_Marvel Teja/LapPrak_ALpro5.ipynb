## 1.1 Latar Belakang

Perulangan merupakan salah satu konsep dasar dalam pemrograman yang digunakan untuk menjalankan suatu perintah secara berulang berdasarkan kondisi atau batas tertentu. ## 2.1 Dasar Teori

Perulangan `for` digunakan untuk menjalankan blok program berulang kali berdasarkan nilai yang dihasilkan oleh suatu iterable, salah satunya `range()`. Bentuk dasar yang digunakan pada latihan adalah `for variabel in range(...):`.

Fungsi `range()` dapat menerima batas awal, batas akhir, dan langkah. Batas akhir tidak termasuk dalam hasil iterasi. Oleh karena itu, bentuk `range(1, n + 1)` digunakan ketika perulangan ingin menghasilkan nilai dari 1 sampai `n`.

Nested `for` adalah penggunaan perulangan `for` di dalam perulangan `for`. Perulangan bagian luar biasanya digunakan untuk menentukan baris, sedangkan perulangan bagian dalam digunakan untuk menentukan banyak karakter atau kolom pada setiap baris.

Parameter `end=""` pada `print()` digunakan agar pencetakan berikutnya tetap berada pada baris yang sama. Setelah satu baris selesai, `print()` tanpa argumen digunakan untuk berpindah ke baris berikutnya.

## 2.2 Kode Program Latihan Sebelumnya

Sebelum mengerjakan tugas utama Pekan 5, terdapat sembilan file latihan yang digunakan untuk memahami perulangan `for`, perulangan bersarang (*nested for*), operasi penjumlahan, seleksi bilangan genap, dan pembentukan pola. Setiap file dijelaskan berdasarkan kode program yang telah dikerjakan.


### 2.2.1 Program `Perulangan_for1_2611532003.py` — Perulangan dasar dengan `range()`


```python
# Buat file dengan nama Perulangan_for1_NIM.py
# Buat program untuk perulangan for dalam Python
# Nama variable ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

ulang_2003 = int(input("Masukkan jumlah perulangan: "))

for i_2003 in range(ulang_2003):
    print(f"Perulangan ke-{i_2003}")
```

**Penjelasan:**

Program menerima jumlah perulangan melalui `input()`, lalu `range(ulang_2003)` membuat urutan mulai dari 0 sampai satu kurang dari batas. Setiap nilai `i_2003` dicetak sebagai nomor perulangan. Variabel `i_2003` berubah otomatis pada setiap iterasi.

**Kaitan dengan tugas Pekan 5:** Konsep utama dari program ini menjadi dasar penyusunan tugas jam pasir, terutama penggunaan `for`, `range()`, `end=""`, kondisi, dan nested `for` untuk mengatur posisi karakter.

### 2.2.2 Program `Perulangan_for2_2611532003.py` — Perbedaan batas awal dan akhir pada `range()`


```python
# Buat file dengan nama Perulangan_for2_NIM.py
# Buat program untuk perulangan for dalam Python
# Nama variable ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

ulang_2003 = int(input("Masukkan jumlah perulangan: "))
print("Perulangan ke-0 sampai ke-", ulang_2003-1)
for i_2003 in range(ulang_2003):
    print(i_2003, end=" ")
print()
print("Perulangan ke-1 sampai ke-", ulang_2003)
for i_2003 in range(1, ulang_2003+1):
    print(i_2003, end=" ")
```

**Penjelasan:**

Program memperlihatkan dua bentuk penggunaan `range()`. Perulangan pertama menggunakan `range(ulang_2003)` sehingga dimulai dari 0 dan berhenti sebelum `ulang_2003`. Perulangan kedua menggunakan `range(1, ulang_2003 + 1)` sehingga dimulai dari 1 dan mencetak sampai nilai batas. `end=" "` digunakan agar angka tetap berada pada baris yang sama.

**Kaitan dengan tugas Pekan 5:** Konsep utama dari program ini menjadi dasar penyusunan tugas jam pasir, terutama penggunaan `for`, `range()`, `end=""`, kondisi, dan nested `for` untuk mengatur posisi karakter.

### 2.2.3 Program `Perulangan_for3_2611532003.py` — Menjumlahkan bilangan menggunakan perulangan `for`


```python
# Buat file dengan nama Perulangan_for1_NIM.py
# Buat program untuk perulangan for dalam Python
# Nama variable ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

ulang_2003 = int(input("Masukkan jumlah perulangan: "))

jumlah_2003 = 0
for i_2003 in range(1, ulang_2003 + 1):
    print(i_2003, end=" ")
    jumlah_2003 = jumlah_2003 + i_2003

    if i_2003 < ulang_2003:
        print(" + ", end="")
    else:
        print(" = ", end="")

print()
print("Jumlah =", jumlah_2003)
```

**Penjelasan:**

Program menghitung jumlah bilangan dari 1 sampai batas. Variabel `jumlah_2003` menjadi akumulator yang ditambah dengan `i_2003` pada setiap iterasi. Kondisi `if` digunakan untuk menampilkan tanda `+` di antara bilangan dan tanda `=` pada bagian akhir.

**Kaitan dengan tugas Pekan 5:** Konsep utama dari program ini menjadi dasar penyusunan tugas jam pasir, terutama penggunaan `for`, `range()`, `end=""`, kondisi, dan nested `for` untuk mengatur posisi karakter.

### 2.2.4 Program `Jumlah_genap_2611532003.py` — Menentukan dan menjumlahkan bilangan genap


```python
# Buat file dengan nama Jumlah_genap_NIM.py
# Buat program untuk perulangan for dalam Python
# Nama variable ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

ulang_2003 = int(input("Masukkan nilai batas: "))

jumlah_2003 = 0
for i_2003 in range(1, ulang_2003 + 1):
    if i_2003 % 2 == 0:
        print(i_2003, end=" ")
        jumlah_2003 = jumlah_2003+ i_2003

        if i_2003 < ulang_2003:
            print("+", end="")
        else:
            print("=", jumlah_2003, end="")
print()
print("Jumlah =", jumlah_2003)
```

**Penjelasan:**

Program menjumlahkan hanya bilangan genap dari 1 sampai batas. Kondisi `i_2003 % 2 == 0` memeriksa apakah bilangan habis dibagi 2. Bilangan yang memenuhi kondisi dicetak dan ditambahkan ke `jumlah_2003`.

**Kaitan dengan tugas Pekan 5:** Konsep utama dari program ini menjadi dasar penyusunan tugas jam pasir, terutama penggunaan `for`, `range()`, `end=""`, kondisi, dan nested `for` untuk mengatur posisi karakter.

### 2.2.5 Program `Latihan5_2611532003.py` — Membentuk pola segitiga dengan perulangan


```python
tinggi_2003 = int(input("Masukkan tinggi segitiga: "))

for i_2003 in range(1, tinggi_2003 + 1):
    print(" " * (tinggi_2003 - i_2003), end="")

    print("* " * i_2003)
```

**Penjelasan:**

Program membentuk pola segitiga berdasarkan tinggi yang dimasukkan. Perulangan luar menentukan nomor baris, sedangkan jumlah spasi dan simbol bintang dihitung dari posisi baris. Program ini menggunakan perkalian string (`" " * ...` dan `"* " * ...`), sehingga menjadi contoh yang perlu dibedakan dari tugas utama Pekan 5 yang melarang perkalian string.

**Kaitan dengan tugas Pekan 5:** Konsep utama dari program ini menjadi dasar penyusunan tugas jam pasir, terutama penggunaan `for`, `range()`, `end=""`, kondisi, dan nested `for` untuk mengatur posisi karakter.

### 2.2.6 Program `Nested_for1_2611532003.py` — Nested `for` untuk mencetak pola titik dan angka


```python
# Buat file dengan nama Nested_for1_NIM.py
# Buat program untuk perulangan for dalam Python
# Nama variable ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

batas_2003 = int(input("Masukkan nilai batas: "))
for line_2003 in range(1, batas_2003 + 1):
    for j_2003 in range(1, (-1 * line_2003 + batas_2003) + 1):
        print(".", end="")
    print(line_2003)
```

**Penjelasan:**

Program menggunakan dua `for` bersarang. Perulangan luar menentukan baris dari 1 sampai batas, sedangkan perulangan dalam mencetak sejumlah titik yang semakin berkurang. Rumus `(-1 * line_2003 + batas_2003) + 1` menghasilkan jumlah titik berdasarkan posisi baris.

**Kaitan dengan tugas Pekan 5:** Konsep utama dari program ini menjadi dasar penyusunan tugas jam pasir, terutama penggunaan `for`, `range()`, `end=""`, kondisi, dan nested `for` untuk mengatur posisi karakter.

### 2.2.7 Program `Nested_for2_2611532003.py` — Nested `for` untuk membentuk persegi bintang


```python
# Buat file dengan nama Nested_for2_NIM.py
# Buat program untuk perulangan for dalam Python
# Nama variable ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

batas_2003 = int(input("Masukkan nilai batas: "))
for i_2003 in range(1, batas_2003 + 1):
    for j_2003 in range(1, batas_2003 + 1):
        print("*", end="")
    print() # Pindah ke baris berikutnya
```

**Penjelasan:**

Program membentuk persegi berisi karakter `*`. Perulangan luar menentukan jumlah baris dan perulangan dalam menentukan jumlah bintang pada setiap baris. Karena kedua batas sama, pola yang terbentuk memiliki jumlah baris dan kolom yang sama.

**Kaitan dengan tugas Pekan 5:** Konsep utama dari program ini menjadi dasar penyusunan tugas jam pasir, terutama penggunaan `for`, `range()`, `end=""`, kondisi, dan nested `for` untuk mengatur posisi karakter.

### 2.2.8 Program `Nested_for3_2611532003.py` — Nested `for` dengan operasi penjumlahan indeks


```python
# Buat file dengan nama Nested_for3_NIM.py
# Buat program untuk perulangan for dalam Python
# Nama variable ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

batas_2003 = int(input("Masukkan nilai batas: "))
for i_2003 in range(batas_2003 + 1):
    for j_2003 in range(batas_2003 + 1):
        print(i_2003 + j_2003, end="")
    print() # Pindah ke baris berikutnya
```

**Penjelasan:**

Program menggunakan indeks `i_2003` dan `j_2003` dari dua perulangan bersarang. Pada setiap posisi dicetak hasil `i_2003 + j_2003`. Karena `range(batas_2003 + 1)` digunakan pada keduanya, indeks berjalan dari 0 sampai batas sehingga terbentuk pola angka berdasarkan jumlah kedua indeks.

**Kaitan dengan tugas Pekan 5:** Konsep utama dari program ini menjadi dasar penyusunan tugas jam pasir, terutama penggunaan `for`, `range()`, `end=""`, kondisi, dan nested `for` untuk mengatur posisi karakter.

### 2.2.9 Program `Nested_for4_2611532003.py` — Nested `for` untuk pola bingkai


```python
# Buat file dengan nama Nested_for4_NIM.py
# Buat program untuk perulangan for dalam Python
# Nama variable ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

tinggi_2003 = int(input("Masukkan tinggi pola (bilangan genap)"))

if tinggi_2003 % 2 !=0:
    print("Tinggi harus bilangan genap")
else:
    a_2003 = tinggi_2003
    c_2003 = a_2003
    lebar_2003 = (2 * tinggi_2003) - 2

    for i_2003 in range(1, tinggi_2003 + 1):
        b_2003 = c_2003 + 1

        for j_2003 in range(1, lebar_2003 + 1):

            # Baris atas dan bawah
            if i_2003 == 1 or i_2003 == tinggi_2003:
                if j_2003 == 1 or j_2003 == lebar_2003:
                    print("#", end="")
                else:
                    print("=", end="")

            #Baris isi
            else:
                if j_2003 == 1 or j_2003 == lebar_2003:
                    print("|", end="")
                else:
                    if j_2003 == c_2003:
                        print("<", end="")
                    elif j_2003 == b_2003:
                        print(">", end="")
                    elif j_2003 == (lebar_2003 - c_2003):
                        print("<", end="")
                    elif j_2003 == (lebar_2003 - c_2003 + 1):
                        print(">", end="")
                    elif j_2003 > b_2003 and j_2003 < (lebar_2003 - c_2003):
                        print(".", end="")
                    else:
                        print(" ", end="")

        print()

        # Logika asli java
        a_2003 -= 2

        if a_2003 <= 0:
            c_2003 = (-a_2003) + 2
        else:
            c_2003 = a_2003
```

**Penjelasan:**

Program membentuk pola dengan bingkai `#`, `=`, dan `|` serta simbol `<>` pada bagian dalam. Program terlebih dahulu memeriksa bahwa `tinggi_2003` merupakan bilangan genap. Perulangan luar mengatur baris, sedangkan perulangan dalam mengatur posisi karakter. Variabel `a_2003` dan `c_2003` digunakan untuk mengubah posisi bagian dalam pola pada setiap baris.

**Kaitan dengan tugas Pekan 5:** Konsep utama dari program ini menjadi dasar penyusunan tugas jam pasir, terutama penggunaan `for`, `range()`, `end=""`, kondisi, dan nested `for` untuk mengatur posisi karakter.

### 2.3.2 Kode Program “tugas5_2003.py”

```python
print("=== PROGRAM JAM PASIR KRISTAL PALINDROMIK (PEKAN 5) ===")
n_2003 = int(input("Masukkan ukuran skala jam pasir (N): "))

if n_2003 <= 0:
    print("Ukuran N harus berupa bilangan bulat positif.")
else:
    lebar_2003 = 4 * n_2003 + 7

    for posisi_2003 in range(1, lebar_2003 + 1):
        if posisi_2003 == 1 or posisi_2003 == lebar_2003:
            print("#", end="")
        else:
            print("=", end="")
    print()

    for baris_2003 in range(n_2003, 0, -1):
        print("|", end="")

        print(" ", end="")

        for spasi_2003 in range(2 * (n_2003 - baris_2003)):
            print(" ", end="")

        for angka_2003 in range(baris_2003, 0, -1):
            print(angka_2003, end="")
            print(" ", end="")

        print("<*>", end="")

        print(" ", end="")
        for angka_2003 in range(1, baris_2003 + 1):
            print(angka_2003, end="")
            if angka_2003 < baris_2003:
                print(" ", end="")

        for spasi_2003 in range(2 * (n_2003 - baris_2003)):
            print(" ", end="")

        print(" ", end="")
        print("|", end="")
        print()

    print("|", end="")
    for spasi_2003 in range(2 * n_2003 + 1):
        print(" ", end="")
    print("<*>", end="")
    for spasi_2003 in range(2 * n_2003 + 1):
        print(" ", end="")
    print("|", end="")
    print()

    for baris_2003 in range(1, n_2003 + 1):
        print("|", end="")

        print(" ", end="")

        for spasi_2003 in range(2 * (n_2003 - baris_2003)):
            print(" ", end="")

        for angka_2003 in range(baris_2003, 0, -1):
            print(angka_2003, end="")
            print(" ", end="")

        print("<*>", end="")

        print(" ", end="")
        for angka_2003 in range(1, baris_2003 + 1):
            print(angka_2003, end="")
            if angka_2003 < baris_2003:
                print(" ", end="")

        for spasi_2003 in range(2 * (n_2003 - baris_2003)):
            print(" ", end="")

        print(" ", end="")
        print("|", end="")
        print()

    for posisi_2003 in range(1, lebar_2003 + 1):
        if posisi_2003 == 1 or posisi_2003 == lebar_2003:
            print("#", end="")
        else:
            print("=", end="")
    print()
```

**Penjelasan:**

5. Fase pertama menggunakan `for baris_2003 in range(n_2003, 0, -1)`, sehingga ukuran deret bergerak dari `N` turun sampai 1.

12. Bingkai bawah dibuat menggunakan nested `for` yang sama dengan bingkai atas.
