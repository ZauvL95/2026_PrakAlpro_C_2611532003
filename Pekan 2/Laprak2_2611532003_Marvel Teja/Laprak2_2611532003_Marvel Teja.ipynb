## 2.1 Dasar Teori

1. Bahasa Pemrograman Python
Python merupakan bahasa pemrograman tingkat tinggi yang bersifat interpreted dan berfokus pada keterbacaan kode. Bahasa ini cocok untuk pemula karena struktur sintaksnya intuitif dan tidak memerlukan deklarasi variabel yang rumit.
2. Tipe data Numerik
Tipe data numerik pada python di bagi menjadi 3:
    a. Integer, Bilangan bulat tanpa mengandung desimal.
    b. Float, Bilangan pecahan yang mengandung angka di belakang koma.
    c. Complex, Bilangan kompleks yang terdiri dari bagian real dan bagian imajiner yang ditandai dengan huruf j, contohnya 100+3j.
3. Tipe data String
Jenis data yang digunakan untuk menyimpan dan mengolah data berbentuk teks, seperti huruf, kata, kalimat, atau simbol.
4. Tipe data Boolean
Tipe data yang hanya memiliki 2 nilai, yaitu True (Benar) dan False (Salah).
5. Syntax input()
Input berfungsi untuk menerima data dari pengguna.
6. Konstanta
Konstanta adalah suatu variabel yang nilainya bersifat tetap dan tidak boleh diubah selama program dijalankan. Python secara bawaan sebenarnya tidak memiliki mekanisme khusus untuk membuat konstanta sejati, karena semua variabel pada dasarnya dapat diubah nilainya. Namun, Python menyediakan modul typing dengan tipe Final yang dapat digunakan sebagai penanda (annotation) bahwa suatu variabel dimaksudkan untuk tidak diubah nilainya.

## 2.2 Kode Program

### 2.2.1 Program "Numerik_2611532003.py"

```python
# Buat file  dengan nama Numerik_NIM.py
# nama variable ditambah 4 digit terakhir NIM contoh: a_5500
# Create a variable with numeric type value.
a_2003 = 100
print("The type of variable having value", a_2003, " is", type(a_2003))

# Create a variable with float value
b_2003 = 10.2345
print("The type of variable having value", b_2003, " is", type(b_2003))

# Create a variable with complex value
c_2003 = 100 + 3j
print("The type of variable having value", c_2003, " is", type(c_2003))
```

**Penjelasan :**
1. Bagian pertama kode ini berfungsi untuk membuat variabel a_2003 dengan nilai 100 dan menampilkan tipe data dari variabel tersebut. Nilai 100 termasuk tipe data integer (int).
2. Bagian kedua berfungsi untuk membuat variabel b_2003 dengan nilai 10.2345 dan menampilkan tipe data dari variabel tersebut. Nilai tersebut termasuk tipe data float karena memiliki angka desimal.
3. Bagian ketiga berfungsi untuk membuat variabel c_2003 dengan nilai 100 + 3j dan menampilkan tipe data dari variabel tersebut. Nilai tersebut termasuk tipe data complex karena memiliki bagian bilangan imajiner yang ditandai dengan j.

### 2.2.2 Program "String_2611532003.py"

```python
# Buat file dengan nama string_NIM.py
# nama variable ditambah 4 digit nim terakhir contoh: nama_1234
# ganti programnya dan data masing2
nama_2003 = "Marvel Teja"
jenis_kelamin_2003 = 'L'
alamat_2003 = """
    Jl. Kotomerapak no 16
    Kec. Padang barat
    Kota padang
"""
agama_2003 = 'Buddha'
print("Nama saya adalah: ", nama_2003)
print("Jenis kelamin saya adalah: ", jenis_kelamin_2003)
print("Alamat saya adalah: ", alamat_2003)
print("Agama saya adalah: ", agama_2003)
```

**Penjelasan :**
1. Bagian pertama kode ini berfungsi untuk membuat variabel nama_2003 yang menyimpan data berupa teks (string), yaitu nama "Marvel Teja".
2. Bagian kedua berfungsi untuk membuat variabel jenis_kelamin_2003 yang menyimpan karakter "L", yang menunjukkan jenis kelamin laki-laki.
3. Bagian ketiga berfungsi untuk membuat variabel alamat_2003 yang menyimpan data teks dalam beberapa baris (multiline string) menggunakan tanda petik tiga ("""). Data tersebut berisi alamat.
4. Bagian keempat berfungsi untuk membuat variabel agama_2003 yang menyimpan data berupa teks, yaitu "Buddha".
5. Bagian kelima berfungsi untuk menampilkan isi dari variabel nama_2003, jenis_kelamin_2003, alamat_2003, dan agama_2003 menggunakan fungsi print().

### 2.2.3 Program "Boolean_2611532003.py"

```python
# Buat file dengan nama Boolean_NIM.py
# Nama variable dtambah 4 digit terakhir contoh: nilai_1234
# Deklarasi variable dengan tipe data boolean
is_lulus_2003 = True
is_cumlaude_2003 = True

# Mengisi Boolean
nilai_2003 = 85
batas_lulus_2003 = 75

# Menentukan nilai Boolean dari kondisi
status_kelulusan_2003 = nilai_2003 >= batas_lulus_2003 # Hasilnya akan True

print("=== Check Kelulusan ===")
print("Nilai:", nilai_2003)
print("Apakah Lulus?:", status_kelulusan_2003)
if is_lulus_2003 and is_cumlaude_2003:
    print("Selamat, Anda lulus dengan predikat Cum Laude")
```

**Penjelasan :**
1. Bagian pertama kode ini berfungsi untuk membuat variabel is_lulus_2003 dan is_cumlaude_2003 dengan nilai True yang menunjukkan kondisi benar.
2. Bagian kedua berfungsi untuk membuat variabel nilai_2003 dengan nilai 85 dan batas_lulus_2003 dengan nilai 75 sebagai nilai minimum kelulusan.
3. Bagian ketiga berfungsi untuk membandingkan nilai_2003 dengan batas_lulus_2003. Karena 85 lebih besar atau sama dengan 75, maka status_kelulusan_2003 menghasilkan nilai True.
4. Bagian keempat berfungsi untuk menampilkan informasi nilai dan status kelulusan menggunakan fungsi print().
5. Bagian kelima menggunakan kondisi if dan operator and untuk memeriksa apakah mahasiswa lulus dan mendapatkan predikat Cum Laude. Karena kedua variabel bernilai True, program menampilkan pesan "Selamat, Anda lulus dengan predikat Cum Laude".

### 2.2.4 Program "Input_2611532003.py"

```python
# Buat file dengan nama Input_NIM.py
# Buat program untuk menghitung luas persegi panjang
# Nama variable ditambah 4 digit nim terakhir contoh: panjang_1234
# Program ini menggunakan fungsi input() untuk meminta pengguna
# memasukkan nilai panjang dan lebar persegi panjang.
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer
panjang_2003 = input('Masukkan nilai panjang: ')
lebar_2003 = input('Masukkan nilai lebar: ')
luas_2003 = int(panjang_2003) * int(lebar_2003)
print("Luas =", luas_2003)
```

**Penjelasan :**
1. Bagian pertama berfungsi untuk meminta pengguna memasukkan nilai panjang menggunakan fungsi input(), kemudian menyimpannya ke dalam variabel panjang_2003.
2. Bagian kedua berfungsi untuk meminta pengguna memasukkan nilai lebar menggunakan fungsi input(), kemudian menyimpannya ke dalam variabel lebar_2003.
3. Bagian ketiga berfungsi untuk menghitung luas persegi panjang dengan mengalikan nilai panjang dan lebar. Fungsi int() digunakan untuk mengubah input yang awalnya berupa teks menjadi bilangan bulat.
4. Bagian keempat berfungsi untuk menampilkan hasil perhitungan luas menggunakan fungsi print().

### 2.2.5 Program "Konstanta_2611532003.py"

```python
# Buat nama dengan nama Konstanta_NIM.py
# Program ini menggunakan konstanta untuk menghitung luas lingkaran
# nama variable ditambah 4 digit nim terakhir contoh: jari_1234

from typing import Final
PI: Final = 3.14
print("pi: %f" % (PI))
jari_2003 = float(input('Masukkan nilai jari jari: '))
luas_2003 = PI * jari_2003 *jari_2003
print("Luas lingkaran dengan jari jari %.2f adalah %.2f" % (jari_2003,luas_2003))
```

**Penjelasan :**
1. Bagian pertama berfungsi untuk mengimpor Final dari modul typing dan menetapkan nilai konstanta PI sebesar 3.14.
2. Bagian kedua berfungsi untuk menampilkan nilai PI dengan format bilangan desimal.
3. Bagian ketiga berfungsi untuk meminta pengguna memasukkan nilai jari-jari lingkaran, kemudian mengubah input tersebut menjadi tipe data float dan menyimpannya dalam variabel jari_2003.
4. Bagian keempat berfungsi untuk menghitung luas lingkaran dengan menggunakan nilai PI dan jari-jari yang telah dimasukkan.
5. Bagian kelima berfungsi untuk menampilkan hasil perhitungan luas lingkaran dengan nilai jari-jari dan luas yang ditampilkan hingga dua angka di belakang koma.
