Git merupakan salah satu sistem version control yang digunakan untuk mencatat perubahan pada suatu proyek. ## 2.1 Dasar Teori
1. Konsep Bahasa Pemrograman Python dan Sintaksis Dasarnya

Python merupakan bahasa pemrograman tingkat tinggi yang mudah dipahami dan digunakan. Python memiliki sintaksis yang sederhana sehingga kode program lebih mudah dibaca dan ditulis. Beberapa sintaksis dasar yang digunakan dalam Python antara lain:

Pencetakan Luaran (Output): Menggunakan fungsi print() untuk menampilkan teks atau data pada layar.
Penggabungan String (String Concatenation): Digunakan untuk menggabungkan dua atau lebih teks menggunakan operator +.
Karakter Escape (Escape Sequence): Digunakan untuk menuliskan karakter khusus dalam sebuah teks dengan menggunakan tanda backslash (\), contohnya \\ untuk menampilkan karakter backslash.
Komentar (Comments): Digunakan untuk memberikan keterangan pada kode program dan tidak akan dijalankan oleh Python. Komentar satu baris diawali dengan tanda #, sedangkan komentar yang terdiri dari beberapa baris dapat menggunakan tiga tanda kutip (""").

2. Lingkungan Kerja Pemrograman (VS Code, Python Debugger, dan Jupyter)

Lingkungan kerja pemrograman membantu pengguna dalam membuat, menjalankan, dan menguji program dengan lebih mudah. Beberapa tools yang digunakan dalam praktikum ini adalah:

Visual Studio Code (VS Code): Editor kode yang digunakan untuk menulis dan menjalankan program. VS Code juga mendukung berbagai bahasa pemrograman dan memiliki banyak fitur tambahan.
Python Debugger: Fitur yang digunakan untuk mencari dan mengetahui kesalahan pada program. Debugger dapat menjalankan program secara bertahap menggunakan breakpoint sehingga kesalahan lebih mudah ditemukan.
Jupyter: Lingkungan pemrograman interaktif yang memungkinkan kode dijalankan dalam beberapa bagian atau cell. Jupyter dapat digunakan untuk mencoba kode secara langsung dan melihat hasilnya dengan lebih mudah.

```python
git init
git add .
git commit -m "Initial commit"
git branch -M main
git remote add origin https://github.com/username/nama-repository.git
git push -u origin main
```

1. git init: Digunakan untuk membuat dan menginisialisasi repositori Git baru pada folder proyek lokal.
2. git add .: Digunakan untuk menambahkan seluruh berkas dan perubahan pada proyek ke dalam staging area.
3. git commit: Digunakan untuk menyimpan riwayat perubahan dari staging area disertai pesan deskriptif.
4. git branch -M main: Digunakan untuk mengubah atau mengonfirmasi nama cabang (branch) utama menjadi main.
5. git remote add origin: Digunakan untuk menghubungkan repositori lokal dengan repositori yang telah dibuat di GitHub.
6. git push: Digunakan untuk mengunggah berkas dan riwayat commit dari komputer lokal ke repositori GitHub.
