## 2.1 Dasar Teori

1. Pengertian Pemrograman Python dan Operator
Python merupakan bahasa pemrograman beraras tinggi (high-level programming language) yang bersifat interpreted, object-oriented, dan memiliki sintaks yang mudah dipahami. Dalam Python, instruksi pemrosesan data dilakukan dengan bantuan operator. Operator adalah simbol-simbol khusus yang digunakan untuk melakukan operasi tertentu terhadap satu atau lebih nilai data yang disebut operand.

2. Operator Aritmatika
Operator aritmatika adalah operator yang digunakan untuk melakukan operasi matematika dasar pada data berjenis numerik (seperti int dan float). Beberapa operator aritmatika utama dalam Python meliputi:
a. Penjumlahan (+): Menambahkan dua buah operand.
b. Pengurangan (-): Mengurangkan operand kanan dari operand kiri.
c. Perkalian (*): Mengalikan dua buah operand.
d. Pembagian (/): Membagi operand kiri dengan operand kanan dan menghasilkan bilangan pecahan (*float*).
e. Pembagian Bulat (//): Membagi operand kiri dengan operand kanan dan membulatkan hasilnya ke bawah (*floor division*).
f. Modulus (%): Mengembalikan sisa hasil pembagian antara dua operand.
g. Perpangkatan (**): Menghitung pangkat dari operand kiri sebesar operand kanan.

3. Operator Perbandingan (Relasional)
Operator perbandingan digunakan untuk membandingkan dua buah nilai. Hasil dari operasi perbandingan selalu bernilai boolean, yaitu True (benar) atau False (salah).
a. Lebih Besar Dari (>) dan Lebih Kecil Dari (<)
b. Lebih Besar Sama Dengan (>=) dan Lebih Kecil Sama Dengan (<=)
c. Sama Dengan (==): Memeriksa apakah nilai kedua operand sama.
d. Tidak Sama Dengan (!=): Memeriksa apakah nilai kedua operand tidak sama.

4. Operator Logika
Operator logika digunakan untuk mengombinasikan dua atau lebih ekspresi relasional atau nilai boolean:
a. and: Menghasilkan True jika seluruh kondisi yang dihubungkan bernilai True.
b. or: Menghasilkan True jika minimal salah satu kondisi bernilai True.
c. not: Membalikkan nilai logika boolean (True menjadi False, dan sebaliknya).

5. Operator Assignment (Penugasan)
Operator assignment digunakan untuk memberikan atau memperbarui nilai ke dalam sebuah variabel. Selain operator dasar penugasan (=), Python menyediakan *compound assignment operators* yang menggabungkan operasi aritmatika dengan penugasan secara langsung, seperti +=, -=, *=, /=, //=, %=, dan **=.

6. Operator Keanggotaan (*Membership Operators*)
Operator keanggotaan digunakan untuk menguji apakah suatu elemen terdapat di dalam sebuah koleksi data (sequence/iterable) seperti string, list, tuple, atau dictionary:
a. in: Bernilai True jika nilai yang dicari ditemukan dalam koleksi data.
b. not in: Bernilai True jika nilai yang dicari tidak ditemukan dalam koleksi data.

7. Operator Identitas (*Identity Operators*)
Operator identitas digunakan untuk membandingkan lokasi memori (address) dari dua buah objek, bukan sekadar nilai isi kontennya:
a. is: Bernilai True jika kedua variabel menunjuk ke objek yang sama di memori RAM.
b. is not: Bernilai True jika kedua variabel menunjuk ke objek yang berbeda di memori RAM.
Perlu dicatat bahwa operator == membandingkan kesamaan nilai (value equality), sedangkan operator is membandingkan kesamaan identitas memori (reference equality).

8. Operator Bitwise
Operator bitwise melakukan operasi pada tingkat bit (biner) dari data numerik integer. Operator ini mengonversi angka desimal menjadi biner terlebih dahulu sebelum melakukan kalkulasi:
a. Bitwise AND (&): Bernilai 1 jika kedua bit bernilai 1.
b. Bitwise OR (|): Bernilai 1 jika salah satu atau kedua bit bernilai 1.
c. Bitwise XOR (^): Bernilai 1 jika kedua bit memiliki nilai yang berbeda.
d. Bitwise NOT (~): Membalikkan seluruh nilai bit (0 menjadi 1, dan 1 menjadi 0).
c. Bitwise Shift Left (<<): Menggeser posisi bit ke kiri sejauh $N$ langkah (mengalikan angka dengan $2^N$).
d. Bitwise Shift Right (>>): Menggeser posisi bit ke kanan sejauh $N$ langkah (membagi angka dengan $2^N$).
Teknik penggabungan dan pengecekan bitwise ini sering dimanfaatkan dalam pemrograman sistem untuk mengelola beberapa indikator kondisi/status hak akses menggunakan satu variabel tunggal (bit masking).

### 2.2.1 Program "Aritmatika_2611532003.py"

```python
# Buat file dengan nama aritmatika_NIM.py
# Buat program untuk operator aritmatika dalam python
# Nama variable ditambah 4 digit nim terakhir contoh: angka1_1234
# Program ini menggunakan fungsi input()
# Nilai yang dimasukkan akan di konversi menjadi tipe data integer

angka1_2003 = int(input("Input angka 1: "))
angka2_2003 = int(input("Input angka 2: "))

#Penjumlahan
hasil_2003 = angka1_2003 + angka2_2003
print("\nOperator Penjumlahan")
print("Hasil =", hasil_2003)

#Pengurangan
hasil_2003 = angka1_2003 - angka2_2003
print("\nOperator Pengurangan")
print("Hasil =", hasil_2003)

#Perkalian
hasil_2003 = angka1_2003 * angka2_2003
print("\nOperator Perkalian")
print("Hasil =", hasil_2003)

#Pembagian
if angka2_2003 != 0:
    hasil_2003 = angka1_2003 / angka2_2003
    print("\nOperator Pembagian")
    print("Hasil =", hasil_2003)

    hasil_2003 = angka1_2003 // angka2_2003
    print("\nOperator Pembagian Bulat")
    print("Hasil =", hasil_2003)

    hasil_2003 = angka1_2003 % angka2_2003
    print("\nOperator Sisa Bagi")
    print("Hasil =", hasil_2003)
else:
    print("Angka kedua tidak boleh bernilai 0")

#Pangkat
hasil_2003 = angka1_2003 ** angka2_2003
print("\nOperator Pangkat")
print("Hasil =", hasil_2003)
```

### 2.2.2 Program "Assignment_2611532003.py"

```python
# Buat file dengan nama assigment_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_1234
# Program ini menngunakan fungsi input()
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer
# Program operator assigment dalam Python

angka1_2003 = int(input("Input angka 1: "))
angka2_2003 = int(input("Input angka 2: "))

print("\nNilai awal angka 1 =",angka1_2003)
print("Nilai awal angka 2 =",angka2_2003)

# Assigment biasa
hasil_2003 = angka1_2003
print("\nAssigment biasa (=)")
print("Hasil =",hasil_2003)

# Assigment penambahan
hasil_2003 = angka1_2003
hasil_2003 += angka2_2003
print("\nAssigment penambahan (+=)")
print("Hasil =",hasil_2003)

# Assigment pengurangan
hasil_2003 = angka1_2003
hasil_2003 -= angka2_2003 
print("\nAssigment pengurangan (-=)")
print("Hasil =",hasil_2003)

# Assigment perkalian
hasil_2003 = angka1_2003
hasil_2003 *= angka2_2003
print("\nAssigment perkalian (*=)")
print("Hasil =",hasil_2003)

# Assigment pembagian, pembagian bulat, dan sisa bagi
if angka2_2003 != 0:
    hasil_2003 = angka1_2003
    hasil_2003 /= angka2_2003
    print("\nAssigment pembagian (/=)")
    print("Hasil =",hasil_2003)
    # Operator tambahan
    hasil_2003 = angka1_2003
    hasil_2003 //= angka2_2003
    print("\nAssigment pembagian bulat (//=)")
    print("Hasil =",hasil_2003)
    hasil_2003 = angka1_2003 
    hasil_2003 %= angka2_2003
    print("\nOperator sisa bagi (%=)")
    print("Hasil =",hasil_2003)
else:
    print("\nPembagian tidak dapat dilakukan.")
    print("Angka kedua tidak boleh bernilai 0.")

# Operator tambahan: assigment perpangkatan
hasil_2003 = angka1_2003
hasil_2003 **= angka2_2003
print("\nOperator perpangkatan (**=)")
print("Hasil =",hasil_2003)
```

**Penjelasan :**
1. Bagian awal berfungsi untuk menerima dua input angka berbentuk integer yang disimpan dalam variabel angka1_2003 dan angka2_2003, lalu menampilkan nilai awalnya.   
2. Bagian assignment biasa (=) berfungsi untuk memberikan atau menyalin nilai dari angka1_2003 langsung ke dalam variabel hasil_2003
3. Bagian assignment aritmatika (+=, -=, *=) berfungsi untuk memodifikasi nilai hasil_2003 secara langsung dengan melakukan penambahan, pengurangan, dan perkalian menggunakan nilai angka2_2003.   
4. Bagian assignment pembagian (/=, //=, %=) mengecek kondisi jika angka2_2003 != 0, lalu mengupdate nilai hasil_2003 dengan hasil pembagian biasa, pembagian bulat, dan sisa bagi secara berturut-turut.   
5. Bagian akhir berfungsi untuk memperbarui nilai hasil_2003 dengan hasil perpangkatan terhadap angka2_2003 menggunakan operator assignment perpangkatan (**=). 

### 2.2.3 Program "Bitwise_2611532003.py"

```python
# Buat file dengan nama Bitwise_NIM.py
# Nama variable ditambah 4 digit nim terakhir contoh: angka1_1234
# Program ini menggunakan fungsi input()

print("===============================")
print("3. OPERATOR BITWISE")
print("===============================")

angka1_2003 = int(input("Input angka bitwise 1: "))
angka2_2003 = int(input("Input angka bitwise 2: "))

print("\nAngka dalam bentuk desimal dan biner")
print("Angka 1", angka1_2003, "| biner =", bin(angka1_2003))
print("Angka 2", angka2_2003, "| biner =", bin(angka2_2003))

# Bitwise AND
hasil_2003 = angka1_2003 & angka2_2003
print("\nBitwise AND (&)")
print(angka1_2003, "&", angka2_2003, "=", hasil_2003)
print("Biner hasil =", bin(hasil_2003))
print("Biner hasil (8 bit) =", format(hasil_2003, "08b"))

# Bitwise OR
hasil_2003 = angka1_2003 | angka2_2003
print("\nBitwise OR (|)")
print(angka1_2003, "|", angka2_2003, "=", hasil_2003)
print("Biner hasil =", bin(hasil_2003))
print("Biner hasil (8 bit) =", format(hasil_2003, "08b"))

# Bitwise XOR
hasil_2003 = angka1_2003 ^ angka2_2003
print("\nBitwise XOR (^)")
print(angka1_2003, "^", angka2_2003, "=", hasil_2003)
print("Biner hasil =", bin(hasil_2003))
print("Biner hasil (8 bit) =", format(hasil_2003, "08b"))

# Bitwise NOT
hasil_2003 = ~angka1_2003
print("\nBitwise NOT (-)")
print("~", angka1_2003, "=", hasil_2003)
print("Biner hasil =", bin(hasil_2003))
print("Biner hasil (8 bit) =", format(hasil_2003, "08b"))

# Bitwise geser kiri
jumlah_geser_2003 = int(input("\nMasukkan jumlah pergeseran bit:"))

hasil_2003 = angka1_2003 << jumlah_geser_2003
print("\nBitwise geser kiri(<<)")
print(angka1_2003, "<<", jumlah_geser_2003,"=", hasil_2003)
print("Biner hasil =", bin(hasil_2003))
print("Biner hasil (8 bit) =", format(hasil_2003, "08b"))

# Bitwise geser kanan
hasil_2003 = angka1_2003 >> jumlah_geser_2003
print("\nBitwise geser kanan(>>)")
print(angka1_2003, ">>", jumlah_geser_2003, "=",hasil_2003)
print("Biner hasil =", bin(hasil_2003))
print("Biner hasil (8 bit) =", format(hasil_2003, "08b"))
```

**Penjelasan :**
1. Bagian awal berfungsi untuk menerima input dua angka desimal, mengonversinya ke integer, lalu menampilkan representasi bentuk biner dari masing-masing angka menggunakan fungsi bin().   
2. Bagian logika bitwise (&, |, ^) berfungsi untuk melakukan operasi biner AND, OR, dan XOR pada tiap bit variabel, kemudian menampilkan hasil angka desimal, biner biasa, serta biner format 8-bit (08b).   
3. Bagian Bitwise NOT (~) berfungsi untuk membalikkan seluruh nilai bit pada variabel angka1_2003 dan menampilkan hasil operasinya.
4. Bagian geser bit (<< dan >>) menerima input jumlah pergeseran bit, lalu menggeser posisi bit dari angka1_2003 ke arah kiri (dikali kelipatan 2) dan ke arah kanan (dibagi kelipatan 2).   

### 2.2.4 Program "Perbandingan_2611532003.py"

```python
# Buat file dengan nama perbandingan_NIM.py
# Nama variable ditambah 4 digit nim terakhir contoh: angka1_1234
# Program ini menggunakan fungsi input()
# Nilai yang dimasukkan akan dikonersikan menjadi tipe data integer
# Program operator perbangingan dalam python

angka1_2003 = int(input("Input angka 1: "))
angka2_2003 = int(input("Input angka 2: "))

# Lebih besar dari
hasil_2003 = angka1_2003 > angka2_2003
print("\nOperator lebih besar dari")
print("Angka 1 > Angka 2 =", hasil_2003)

# Lebih kecil dari
hasil_2003 = angka1_2003 < angka2_2003
print("\nOperator lebih kecil dari")
print("Angka 1 < Angka 2 =", hasil_2003)

# Lebih besar dari atau sama dengan
hasil_2003 = angka1_2003 >= angka2_2003
print("\nOperator lebih besar dari atau sama dengan")
print("Angka 1 >= Angka 2 =", hasil_2003)

# Lebih kecil dari atau sama dengan
hasil_2003 = angka1_2003 <= angka2_2003
print("\nOperator lebih kecil dari atau sama dengan")
print("Angka 1 <= Angka 2 =", hasil_2003)

# Sama dengan
hasil_2003 = angka1_2003 == angka2_2003
print("\nOperator sama dengan")
print("Angka 1 == Angka 2 =", hasil_2003)

# Tidak sama dengan
hasil_2003 = angka1_2003 != angka2_2003
print("\nOperator tidak sama dengan")
print("Angka 1 != Angka 2 =", hasil_2003)

# Tambahan: perbandingan berantai dalam Python
hasil_2003 = 0 < angka1_2003 < 100
print("\nPerbandingan berantai")
print("@ < Angka 1 < 100 =", hasil_2003)

hasil_2003 = 0 < angka2_2003 < 100
print("0 < Angka 2 < 100 =", hasil_2003)
```

**Penjelasan :**
1. Bagian awal berfungsi untuk menerima input dua angka berjenis integer yang disimpan dalam variabel angka1_2003 dan angka2_2003.
2. Bagian relasional (>, <, >=, <=, ==, !=) berfungsi untuk membandingkan hubungan antara angka1_2003 dan angka2_2003, yang akan menghasilkan nilai boolean berupa True atau False.   
3. Bagian perbandingan berantai berfungsi untuk memeriksa apakah nilai variabel berada dalam rentang tertentu secara sekaligus (memeriksa apakah angka bernilai lebih besar dari 0 dan kurang dari 100).

### 2.2.5 Program "Logika_2611532003.py"

```python
# Buat file dengan nama logika_NIM.py
# Nama variable ditambah 4 digit angka terakhir contoh:a1_1234
# Program ini mengggunakan fungsi input()
# Program operator logika dalam python

# Masukkan nilai boolean
# Input tidak peka terhadap huruf besar dan kecil
a1_2003 = input("Input nilai boolean 1 (true/false)").strip().lower() == "true"
a2_2003 = input("Input nilai boolean 2 (true/false)").strip().lower() == "true"

print("\nA1 =",a1_2003)
print("A2", a2_2003)

# Konjungsi: bernilai True jika keduanya True
hasil_2003 = a1_2003 and a2_2003
print("\nKonjungsi (AND)")
print("A1 and A2 =", hasil_2003)

# Disjungsi: bernilai True jika salah satunya True
hasil_2003 = a1_2003 or a2_2003
print("\nKonjungsi (OR)")
print("A1 or A2 =", hasil_2003)

# Negasi A1: membalik nilai A1
hasil_2003 = not a1_2003
print("\nNegasi A1 (NOT)")
print("not A1 =", hasil_2003)

# Negasi A2: membalik nilai A1
hasil_2003 = not a2_2003
print("\nNegasi A12 (NOT)")
print("not A2 =", hasil_2003)

# XOR: Bernilai True jika kedua nilai berbeda
hasil_2003 = a1_2003 != a2_2003
print("\nDisjungsi Eksklusif (XOR)")
print("A1 XOR A2 =", hasil_2003)
```

**Penjelasan :**

2. Bagian Konjungsi (and) dan Disjungsi (or) mengevaluasi dua variabel boolean; and bernilai True jika keduanya True, sedangkan or bernilai True jika salah satunya True.   
3. Bagian Negasi (not) berfungsi untuk membalikkan nilai boolean dari variabel, mengubah True menjadi False dan sebaliknya.

1. Jelaskan operator aritmatika apa saja yang digunakan dan bagaimana operator tersebut membantu menghitung transaksi.

Operator pengurangan (-) digunakan untuk menghitung total pembayaran setelah dikurangi diskon. Operator pembagian (/) digunakan untuk menghitung rata-rata harga barang. Operator modulus (%) digunakan untuk mencari sisa pembagian jumlah barang dengan angka 2. Sementara itu, operator += digunakan untuk menambahkan poin pelanggan. Operator-operator tersebut membantu program melakukan proses perhitungan transaksi secara otomatis.

Operator and digunakan ketika semua kondisi harus terpenuhi. Contohnya, diskon_member_2003 bernilai True apabila pelanggan merupakan member dan memenuhi syarat minimal belanja Rp200.000. Operator or digunakan apabila cukup salah satu kondisi yang terpenuhi, seperti pada akses_khusus_2003, yaitu pelanggan mendapatkan akses khusus jika memenuhi akses member atau akses promo. 4. Apa perbedaan antara is dan == pada Python?
Operator == digunakan untuk membandingkan nilai atau isi dari dua objek, sedangkan operator is digunakan untuk memeriksa apakah dua variabel merujuk pada objek yang sama di memori. Pada program, nilai_a_2003 is nilai_b_2003 digunakan untuk memeriksa identitas objek. Jadi, is bukan pengganti == untuk membandingkan nilai. Untuk membandingkan isi atau nilai, digunakan ==.

Operator OR (|) digunakan untuk menggabungkan kondisi-kondisi tersebut ke dalam kode_status_2003. Selanjutnya, operator AND (&) digunakan untuk memeriksa kondisi tertentu.
